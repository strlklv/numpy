#### 35. How to compute ((A+B)*(-A/2)) in place (without copy)? (★★☆)

````python
import numpy as np

A = np.random.random(5)
B = np.random.random(5)

np.add(A, B, out=A)          # шаг 1: A = A + B
np.divide(A, -2, out=B)      # шаг 2: B = -A/2
np.multiply(A, B, out=A)     # шаг 3: A = A * B

print(A)
````
